## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def main():

    print("A continuación se le pedira que ingrese el número de mes a consultar el crecimiento del vivero:")

    dig = {}

    while True:

        decision = int(input("\n¿Desea continuar? (1)Sí (2)No:"))

        while decision != 1 and decision != 2:

            decision = int(input("\nIngrese una opción numérica válida.\n¿Desea continuar? (1)Sí (2)No:"))

        if decision == 1:

            sucesion = int(input("\nIngrese el mes a consultar el crecimiento de producción en el vivero:"))

            if sucesion in dig:

                print(f"Por medio de memoización, la producción en el mes #{sucesion} es de {dig[sucesion]} nuevas plantas.")

            else:

                dig[sucesion] = fibonacci(sucesion)

                print(f"La producción en el mes #{sucesion} es de {dig[sucesion]} plantas nuevas.")

        elif decision == 2:

            break



def fibonacci(mes):

    if mes == 0:

        return 0
    
    elif mes == 1:

        return 1

    elif mes >= 2:

        return fibonacci(mes-1) + fibonacci(mes-2) 

main() 
```
