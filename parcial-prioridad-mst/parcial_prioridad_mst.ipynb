```python
import math
import matplotlib.pyplot as plt

# ==============================================================================
# PARTE 1 Y 2: MAX-PRIORITY QUEUE CON MAX-HEAP
# ==============================================================================

class Tarea:
    def __init__(self, codigo, descripcion, prioridad):
        self.codigo = codigo
        self.descripcion = descripcion
        self.prioridad = prioridad

    def __repr__(self):
        return f"[{self.codigo}: {self.descripcion} (P:{self.prioridad})]"


def max_heapify(arr, n, i):
    mayor = i
    izq = 2 * i + 1
    der = 2 * i + 2

    if izq < n and arr[izq].prioridad > arr[mayor].prioridad:
        mayor = izq

    if der < n and arr[der].prioridad > arr[mayor].prioridad:
        mayor = der

    if mayor != i:
        arr[i], arr[mayor] = arr[mayor], arr[i]
        max_heapify(arr, n, mayor)


def build_max_heap(arr):
    n = len(arr)
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(arr, n, i)


def extract_max(arr):
    if len(arr) == 0:
        return None
    max_elem = arr[0]
    arr[0] = arr[-1]
    arr.pop()
    if len(arr) > 0:
        max_heapify(arr, len(arr), 0)
    return max_elem


def imprimir_arbol(arr, i=0, nivel=0, prefijo="Raíz: "):
    if i < len(arr):
        print(" " * (nivel * 4) + prefijo + f"{arr[i].codigo} (P:{arr[i].prioridad})")
        imprimir_arbol(arr, 2 * i + 1, nivel + 1, "Izq--- ")
        imprimir_arbol(arr, 2 * i + 2, nivel + 1, "Der--- ")


# Datos iniciales
tareas = [
    Tarea("T01", "configurar servidor", 6),
    Tarea("T02", "reparar sensor", 10),
    Tarea("T03", "actualizar software", 4),
    Tarea("T04", "revisar cámara", 8),
    Tarea("T05", "generar respaldo", 3),
    Tarea("T06", "configurar alarma", 7),
]

print("=" * 60)
print("1. ARREGLO INICIAL:")
for t in tareas:
    print("  ", t)

build_max_heap(tareas)

print("\n2. MAX-HEAP RESULTANTE (ARREGLO):")
for t in tareas:
    print("  ", t)

print("\nÁRBOL CORRESPONDIENTE AL MAX-HEAP:")
imprimir_arbol(tareas)

tarea_extraida = extract_max(tareas)
print(f"\nEXTRACT-MAX REALIZADO -> Tarea atendida: {tarea_extraida}")

print("\nMAX-HEAP DESPUÉS DE EXTRACT-MAX:")
for t in tareas:
    print("  ", t)

print("\nÁRBOL ACTUALIZADO:")
imprimir_arbol(tareas)


# ==============================================================================
# PARTE 3 Y 4: DISTANCIAS, GRAFO Y ALGORITMO DE KRUSKAL
# ==============================================================================

# Coordenadas (x, y) de los 6 dispositivos (puedes ajustar estos valores si tienes otros)
dispositivos = {
    "D1 (Servidor)": (2, 8),
    "D2 (Sensor)": (5, 9),
    "D3 (PC Software)": (8, 7),
    "D4 (Cámara)": (3, 3),
    "D5 (Respaldo)": (7, 2),
    "D6 (Alarma)": (9, 4),
}

nombres = list(dispositivos.keys())
aristas = []

print("\n" + "=" * 60)
print("3. DISTANCIAS ENTRE CADA PAR DE DISPOSITIVOS (ARISTAS PONDERADAS):")
for i in range(len(nombres)):
    for j in range(i + 1, len(nombres)):
        u, v = nombres[i], nombres[j]
        x1, y1 = dispositivos[u]
        x2, y2 = dispositivos[v]
        distancia = round(math.hypot(x2 - x1, y2 - y1), 2)
        aristas.append((distancia, u, v))
        print(f"  {u} <---> {v} : {distancia} m")


# Estructura Disjoint Set (Union-Find) para Kruskal
class UnionFind:
    def __init__(self, nodos):
        self.padre = {nodo: nodo for nodo in nodos}
        self.rango = {nodo: 0 for nodo in nodos}

    def find(self, nodo):
        if self.padre[nodo] != nodo:
            self.padre[nodo] = self.find(self.padre[nodo])
        return self.padre[nodo]

    def union(self, u, v):
        raiz_u = self.find(u)
        raiz_v = self.find(v)
        if raiz_u != raiz_v:
            if self.rango[raiz_u] < self.rango[raiz_v]:
                self.padre[raiz_u] = raiz_v
            elif self.rango[raiz_u] > self.rango[raiz_v]:
                self.padre[raiz_v] = raiz_u
            else:
                self.padre[raiz_v] = raiz_u
                self.rango[raiz_u] += 1
            return True
        return False


def kruskal(nodos, lista_aristas):
    # Ordenar aristas por peso (distancia) ascendente
    aristas_ordenadas = sorted(lista_aristas, key=lambda x: x[0])
    uf = UnionFind(nodos)
    mst = []
    costo_total = 0.0

    for peso, u, v in aristas_ordenadas:
        if uf.union(u, v):
            mst.append((u, v, peso))
            costo_total += peso

    return mst, round(costo_total, 2)


mst_resultado, cable_total = kruskal(nombres, aristas)

print("\n" + "=" * 60)
print("4. ALGORITMO DE KRUSKAL (ÁRBOL DE EXPANSIÓN MÍNIMA):")
for u, v, peso in mst_resultado:
    print(f"  Conectar {u} -- {v} | Cable: {peso} m")
print(f"\nCANTIDAD MÍNIMA TOTAL DE CABLE: {cable_total} m")


# ==============================================================================
# REPRESENTACIÓN GRÁFICA (GRAFO COMPLETO VS KRUSKAL MST)
# ==============================================================================

plt.figure(figsize=(10, 6))

# Dibujar todas las posibles conexiones en gris punteado
for peso, u, v in aristas:
    x_vals = [dispositivos[u][0], dispositivos[v][0]]
    y_vals = [dispositivos[u][1], dispositivos[v][1]]
    plt.plot(x_vals, y_vals, color="lightgray", linestyle="--", linewidth=1, zorder=1)

# Dibujar las conexiones seleccionadas por Kruskal en verde sólido
for u, v, peso in mst_resultado:
    x_vals = [dispositivos[u][0], dispositivos[v][0]]
    y_vals = [dispositivos[u][1], dispositivos[v][1]]
    plt.plot(x_vals, y_vals, color="#2ecc71", linewidth=3, zorder=2, label="Cableado óptimo (Kruskal)" if (u, v, peso) == mst_resultado[0] else "")
    
    # Etiqueta con el peso en el punto medio
    mid_x = (x_vals[0] + x_vals[1]) / 2
    mid_y = (y_vals[0] + y_vals[1]) / 2
    plt.text(mid_x, mid_y, f"{peso}m", color="darkgreen", fontweight="bold", fontsize=9,
             bbox=dict(boxstyle="round,pad=0.2", facecolor="white", edgecolor="#2ecc71", alpha=0.9))

# Dibujar los nodos (dispositivos)
for nombre, (x, y) in dispositivos.items():
    plt.scatter(x, y, color="#2980b9", s=200, zorder=3)
    plt.text(x, y + 0.3, nombre, fontsize=10, fontweight="bold", ha="center", color="#1a252f")

plt.title(f"Red de Dispositivos - Algoritmo de Kruskal (Cable total: {cable_total} m)", fontsize=13, fontweight="bold")
plt.xlabel("Coordenada X (metros)")
plt.ylabel("Coordenada Y (metros)")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
